```python
import scipy
import numpy as np
import matplotlib.pyplot as plt

```

```python
np.random.seed(0)

# generate a normal distribution 1D data
norm_dis_1D_data = np.random.normal(loc=0, scale=1, size=500)

# generate a normal distribution 2D data
norm_dis_2D_data_x = np.random.normal(0, 1, 300)
norm_dis_2D_data_y = np.random.normal(0, 1, 300)
norm_dis_2D_data = np.vstack([norm_dis_2D_data_x, norm_dis_2D_data_y])
```

```python
# gaussian_kde KDE : Kernel Density Estimation
    # 1D 
kde = scipy.stats.gaussian_kde(norm_dis_1D_data)
    # bw_method : setting the gaussian kernel width (usuall default)

x = np.linspace(min(norm_dis_1D_data), max(norm_dis_1D_data), 100)
# -4, 4 is the showing range (almost shows all the density)
# x = np.linspace(-4, 4, 400)

# estimate density
density = kde(x)

plt.hist(norm_dis_1D_data, bins=30, density=True, alpha=0.4)
plt.plot(x, density)
plt.show()
```

Kernel Density Estimation
    able to check whether the issue is bias(under-fit) or ()
    if the line shift to left or right (based on 0), then the model might be bias

```python
# gaussian_kde KDE : Kernel Density Estimation
    # 2D 
kde = scipy.stats.gaussian_kde(norm_dis_2D_data)

xx, yy = np.mgrid[-4:4:100j, -4:4:100j]
grid = np.vstack([xx.ravel(), yy.ravel()])

# estimate density
zz = kde(grid).reshape(xx.shape)

plt.contourf(xx, yy, zz)
plt.scatter(norm_dis_2D_data_x, norm_dis_2D_data_y, s=5)
plt.show()
```
